## Passeio Aleatório Unidimensional Simétrico

### Suporte Teórico

Sejam $N$ $\in$ $\mathbb{N}$ fixado e $\mathcal{F}$ a família de sequências $\left\{ S_k \right\}_{k = 0}^{N}$ dada por: 

$$ S_0 = 0 .$$
$$ S_k = S_{k-1} + x_k, x_k \in \{-1,1\}, \forall k \geq 1 .$$

Seja $\mathcal{G} = \{-1,1\}^N$ a família de vetores $(x_1, x_2, ..., x_N)$.

Considere o seguinte racicínio.
* $\forall \omega \in \mathcal{F}$, $\exists \omega' \in \mathcal{G}$ tal que a aplicação de $\omega'$ na lei de construção de $\mathcal{F}$ retorna $\omega$.

* Supondo que $\exists \omega \in \mathcal{F}$ tal que mais um $\omega' \in \mathcal{G}$ satisfaz a aplicação anterior, se a aplicação dos vetores $v_1, v_2, ..., v_m$ retorna $\omega$, então
quaisquer desses vetores tomados dois a dois coincidem em todas as posições. Daí que, na realidade, eles são o mesmo vetor, ou seja, o $\omega'$ é único.

Diante disso e sabendo que todo $\omega' \in \mathcal{G}$ é aplicável na lei de $\mathcal{F}$, vem que $\mathcal{F}$ e $\mathcal{G}$ estabalecem uma **relação biunívoca** entre seus elementos.

### Passeio Aleatório Unidimensional Simétrico (PADS)

Seja $\mathscr{E}$ o experimento no qual uma sequência qualquer de $\mathcal{F}$ será montada de modo que, em cada $k \geq 1$, o valor de $x_k$ seja aleatoriamente selecionado de $\{-1,1\}.$
Nesse sentido, temos:

* True $\equiv \Omega = \mathcal{F} \equiv \Omega = \mathcal{G}$.
* Por definição, $\Omega$ equiprovável com $P\left[\omega\right] = \left( 1/2 \right)^N, \forall \omega \in \Omega, $ e $P\left[E\right] = \left( \#E/ \# \Omega\right), \forall E \subset \Omega$.

Ademais, sejam $X_1$, $X_2$, $...$, $X_N$ as variáveis aleatórias que assumem o valor de $x_k, k \in \{1, 2, ..., N\}$.
Daí:

* $ \Omega _{X_k} = \{-1,1\} $.

* $P\left[ X_k = -1 \right] = P\left[ X_k = 1 \right] =  1/2 $.

Diante disso, definimos **o Passeio Aleatório Unidimensional Simétrico** como o experimento $\mathscr{E}$ com as variáveis aleatórias $X_k$. Particularmente para referenciar um $\omega \in \Omega$, diremos "**um** passeio aleatório (unidimensional simétrico)".

### Situação-Problema

Considerando a execução do PADS, qual a probabilidade do agente finalizar o passeio na origem?

Seja $E$ o evento "passeio finaliza na origem". Para que $E$ ocorra, é necessário e suficiente que $S_N = 0$. Para que $S_N = 0$, é necessário e suficiente que a quantidade de passos $+1$ seja igual à quantidade de passos $-1$. Deste modo, temos que:

* Se $N$ é ímpar, então $P\left[E\right] = 0$.

* Se $N$ é par, então:

$$
\begin{aligned}
P\left[E\right] &= P\left[S_N = 0\right] \\[1ex]
&= \dfrac{\dbinom{N}{N/2}}{2^N}
\end{aligned}
$$

## Teorema do Limite Central

### Suporte Teórico

#### 1. O Teorema do Limite Central
Segundo o Teorema do Limite Central (TCL), a média de amostras variáveis aleatórias independentes de uma mesma distribuição (que possua média e variância finitas) tende a uma distribuição normal, mesmo em casos de distribuições assimétricas, como a exponencial. Ou seja, as médias das amostras selecionadas aleatoriamente com base na distribuição previamente escolhida irão tender à equação da distribuição normal:

$$ f(x) = \frac{1}{\sigma\sqrt{2\pi}}\, e^{-\frac{(x-\mu)^2}{2\sigma^2}} $$

#### 2. A Distribuição Exponencial
Para contextualizar, a distribuição exponencial consiste em uma função de probabilidade assimétrica que decai com o tempo, o que indica que maiores valores da variável aleatória tem menor probabilidade de ocorrer. Tem-se a seguinte notação:

$$ X \sim Exp(\lambda), \quad \forall \lambda > 0. $$

Em que $\lambda$ é a taxa de ocorrência do evento em um período de tempo. Sua função de probabilidade é dada por

$$ f(x) = \lambda e ^{-\lambda x}, \quad \forall x \geq 0. $$

A esperança teórica, ou seja, a taxa média de tempo entre a ocorrência e cada evento é dada por

$$ E(X) = \frac{1}{\lambda}. $$

Sendo assim, sua variânia teórica também pode ser dada por

$$ Var(X) = \frac{1}{\lambda^{2}}. $$

O que indica, portanto, que seu desvio padrão é igual à média

$$ \sigma = \mu = \frac{1}{\lambda}. $$

### O Experimento Realizado

Diante disso, com base em uma variável aleatória exponencial, com o parâmetro lambda inserido pelo usuário, utiliza-se a biblioteca `numpy` para selecionar aleatoriamente uma amostra de escolha do usuário, sendo esse experimento repetido, também, o número de vezes que o usuário solicitar, sendo, assim, formada uma matriz em que o número de linhas é o número de repetições solicitadas, e o número de colunas é o tamanho das amostras.

Cada amostra possuirá uma média, o que gera uma lista de médias, e a média dessa lista deve se aproximar da esperança teórica. Por fim, os resultados, como solicitados, são escritos em um arquivo `.csv`. e o histograma que mostra a aproximação com a normal é plotado.
